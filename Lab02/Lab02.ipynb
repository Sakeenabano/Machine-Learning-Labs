## Task 3 · Randomness and reproducibility
- **(a)** Set `np.random.seed(2026)`, then generate one random number between 0 and 1.
- **(b)** Generate 25 numbers from the standard normal distribution.
- **(c)** Set `np.random.seed(42)` and roll 10,000 Ludo dice with `np.random.randint`. Count each face with `np.unique(..., return_counts=True)`.
- **(d)** Estimate P(six) and compare it with 1/6.
- **(e)** Count how many times two sixes came in a row (no loops).

```python
# T3A - Generate one random value using seed 2026
np.random.seed(2026)
random_value = np.random.rand()
print("T3A - Random Value:", random_value)

# T3B - Generate 25 normally distributed values
normal_values = np.random.randn(25)
print("T3B - 25 Normal Values:\n", normal_values)

# T3C - Simulate 10,000 pairs of Ludo dice rolls
np.random.seed(42)
dice = np.random.randint(1, 7, size=(10000, 2))
print("T3C - First 10 Dice Rolls:\n", dice[:10])

# T3D - Count occurrences of each dice face
faces, counts = np.unique(dice, return_counts=True)
print("T3D - Face Counts:")
print(dict(zip(faces, counts)))

# T3E - Calculate probability of getting a six
probability_six = np.sum(dice == 6) / dice.size
print("T3E - Probability of Six:", probability_six)

# Count double sixes
double_sixes = np.sum(np.all(dice == 6, axis=1))
print("T3E - Number of Double Sixes:", double_sixes)
```

## Task 8 · Inflation FY2024 vs FY2025 (real data, Pakistan Economic Survey 2024-25)
`cpi` has 2 rows (FY2024, FY2025) and 10 columns (Jul to Apr), year-on-year CPI inflation in %.
- **(a)** Average inflation of each year.

- **(c)** Drop in inflation for each month (FY2024 − FY2025).
- **(d)** The month with the biggest drop.
- **(e)** Label each FY2025 month "Low" (< 5), "Moderate" (5–10) or "High" (≥ 10) with `np.where`.

**Interpret:** in 1–2 lines, what happened to prices in Pakistan between the two years?

```python
# Task 8: Inflation FY2024 vs FY2025

# A. Average inflation for each year
average_inflation = np.mean(cpi, axis=1)
print("T8A - Average Inflation (%):")
print("FY2024:", round(average_inflation[0], 2), "%")
print("FY2025:", round(average_inflation[1], 2), "%")

# B. FY2024 months above 25%
high_months = cpi_months[cpi[0] > 25]
print("\nT8B - FY2024 Months Above 25%:")
print(high_months)

# C. Inflation drop per month
inflation_drop = cpi[0] - cpi[1]
print("\nT8C - Monthly Inflation Drop (% points):")
print(np.round(inflation_drop, 2))

# D. Month with biggest drop
biggest_drop_index = np.argmax(inflation_drop)
print("\nT8D - Biggest Drop Month:", cpi_months[biggest_drop_index])
print("Biggest Drop:", round(inflation_drop[biggest_drop_index], 2), "percentage points")

# E. Label FY2025 inflation
inflation_labels = np.where(
    cpi[1] < 5, "Low",
    np.where(cpi[1] < 10, "Moderate", "High")
)

print("\nT8E - FY2025 Inflation Labels:")
print(inflation_labels)
```


### Interpretation

The average inflation decreased from 26.22% in FY2024 to 4.87% in FY2025. January had the biggest decrease of 25.9 percentage points. This shows that prices were increasing much more slowly in FY2025. However, lower inflation does not mean that prices became cheaper; it means the rate of price increase decreased.

## Task 9 · Grace marks
Using `marks`:
- **(a)** Add 5 grace marks but cap at 100 (`np.clip`).
- **(b)** Pass % (≥ 50) after grace.
- **(c)** Index of students who passed **only because of** the grace marks.
- **(d)** z-score of each student after grace.

```python
# T9A - Add 5 grace marks and cap marks at 100
grace_marks = np.clip(marks + 5, 0, 100)

print("T9A - Original Marks:", marks)
print("T9A - Marks After Grace:", grace_marks)

# T9B - Calculate pass percentage after grace marks
passed = grace_marks >= 50
pass_percentage = np.mean(passed) * 100

print("\nT9B - Pass Percentage After Grace:", pass_percentage, "%")

# T9C - Find students who passed only because of grace marks
passed_only_by_grace = np.where((marks < 50) & (grace_marks >= 50))[0]

print("\nT9C - Students Passed Only Because of Grace Marks:")
print(passed_only_by_grace)

# T9D - Calculate Z-scores after grace marks
mean_marks = np.mean(grace_marks)
std_marks = np.std(grace_marks)

z_scores = (grace_marks - mean_marks) / std_marks

print("\nT9D - Z-Scores After Grace:")
print(np.round(z_scores, 2))
```

```python
# T10A - Create an 8x8 image and draw number 7
img = np.zeros((8, 8), dtype=int)

img[1, 1:7] = 255
img[2:7, 5] = 255
img[4, 3:7] = 128

print("T10A - Original Image:")
print(img)

# T10B - Flip image left to right
flipped_img = img[:, ::-1]
print("\nT10B - Flipped Image:")
print(flipped_img)

# T10C - Invert the image
inverted_img = 255 - img
print("\nT10C - Inverted Image:")
print(inverted_img)

# T10D - Crop rows 1-4 and columns 3-6
cropped_img = img[1:5, 3:7]
print("\nT10D - Cropped Image:")
print(cropped_img)

# T10E - Brighten image by 80, capped at 255
bright_img = np.clip(img + 80, 0, 255)
print("\nT10E - Brightened Image:")
print(bright_img)

# T10F - Image information
print("\nT10F - Image Information:")
print("Shape:", img.shape)
print("Data Type:", img.dtype)
print("Number of White Pixels:", np.sum(img == 255))
```
